#### Refined SAM3 model with clip (True to docmentation)

```python
import os
from pathlib import Path
from ultralytics.models.sam import SAM3SemanticPredictor

# Define directories for your dataset
img_dir = Path("../data/sam_data/damage")
txt_dir = Path("../data/sam_data/YOLO_Labels")
vis_dir = Path("../data/sam_data/Visuals")

txt_dir.mkdir(parents=True, exist_ok=True)
vis_dir.mkdir(parents=True, exist_ok=True)

# List of valid image extensions to prevent silent skipping
valid_extensions = [".jpg", ".jpeg", ".png", ".webp"]

# Explicit descriptive phrases for damages only.
# The index of each string exactly matches your master_mapper (0 to 7).
damage_queries = [
    "missing car part",         # 0
    "broken car part",          # 1
    "scratch on car body",      # 2
    "cracked car window",       # 3
    "dent on car body",         # 4
    "flaking car paint",        # 5
    "car paint chip",           # 6
    "rust and corrosion on car" # 7
]

print("Initializing SAM3SemanticPredictor with configuration...")

# Initialize predictor with configuration overrides based on the official template
overrides = dict(
    conf=0.25,
    task="segment",
    mode="predict",
    model="sam3.pt",
    half=False,    # Set to False for CPU stability (True is for CUDA FP16)
    device='cuda',  # Forcing CPU to avoid AMD ROCm driver issues
    save=False,    # Disabled auto-save so we can manually route files to specific folders
)

predictor = SAM3SemanticPredictor(overrides=overrides)

print(f"Processing images in {img_dir}...")

for img_path in img_dir.rglob("*"):
    if img_path.suffix.lower() not in valid_extensions:
        continue
        
    print(f"Processing: {img_path.name}")
    
    try:
        # Set image once for multiple queries as per the template
        predictor.set_image(str(img_path))
        
        # Query with multiple text prompts (list of strings)
        results = predictor(text=damage_queries)
        
        if not results or len(results) == 0:
            continue
            
        # The predictor returns a list, grab the first (and only) result object
        res = results[0]
        
        # Save visualizer image to inspect accuracy manually
        res.save(filename=str(vis_dir / f"{img_path.stem}_visual.jpg"))
        
        # Save YOLO formatted TXT labels
        if res.masks is not None and len(res.masks) > 0:
            with open(txt_dir / f"{img_path.stem}.txt", "w") as f:
                class_ids = res.boxes.cls.tolist()
                
                for cls_id, polygon in zip(class_ids, res.masks.xyn):
                    # Flatten the polygon coordinates
                    flat_coords = " ".join([f"{p[0]:.6f} {p[1]:.6f}" for p in polygon])
                    
                    # Write class_id (0-7 based on damage_queries index) and coordinates
                    f.write(f"{int(cls_id)} {flat_coords}\n")
                    
    except Exception as e:
        print(f"Error processing {img_path.name}: {e}")

print("Done! Please check the Visuals folder to verify SAM 3 accuracy with CLIP installed.")
```

#### Generic SAM3 model

```python
import os
from pathlib import Path
from ultralytics import SAM
from ultralytics import YOLO

# 1. قائمة الكلاسات مدمجة كنص واحد (يفصل بينها نقطة ليفهمها SAM 3 كـ Concepts)
# الترتيب هنا من 0 إلى 28 ليطابق master_mapper الخاص بك تماماً
text_prompt = "missing car part . broken car part . scratch on car body . cracked car window . dent on car body . flaking car paint . car paint chip . rust and corrosion on car"
# 2. إعداد المجلدات
img_dir = Path("../data/sam_data/damage")
txt_dir = Path("../data/sam_data/YOLO_Labels")
vis_dir = Path("../data/sam_data/Visuals")

txt_dir.mkdir(parents=True, exist_ok=True)
vis_dir.mkdir(parents=True, exist_ok=True)

# 3. تحميل بطل الليلة (SAM 3)
print("Loading SAM 3 Model...")
# تأكد أن ملف sam3.pt موجود في نفس المجلد
model = SAM("sam3.pt") 

# 4. بدء المعالجة
print(f"Processing images in {img_dir}...")
# حل مشكلة التجاهل الصامت: أضفنا webp
valid_extensions = [".jpg", ".jpeg", ".png", ".webp"] 

# نستخدم rglob للبحث حتى داخل المجلدات الفرعية لو وجدت
for img_path in img_dir.rglob("*"):
    if img_path.suffix.lower() not in valid_extensions:
        continue
        
    print(f"Processing: {img_path.name}")
    
    # SAM 3 يقوم بالاكتشاف والتقطيع معاً بناءً على النص فقط!
    results = model.predict(
        source=img_path, 
        labels=text_prompt, 
        conf=0.25, 
        device='cuda', # استخدم cpu مؤقتاً لتجنب أي مشاكل مع كرت AMD
        verbose=False
    )[0]
    
    # حفظ الصورة المرسومة للتأكد بالعين
    results.save(filename=str(vis_dir / f"{img_path.stem}_visual.jpg"))
    
    # حفظ ملف الـ TXT بتنسيق Segmentation
    if results.masks:
        with open(txt_dir / f"{img_path.stem}.txt", "w") as f:
            # SAM 3 يرجع رقم الكلاس بناءً على ترتيبه في text_prompt
            class_ids = results.boxes.cls.tolist() 
            for cls_id, polygon in zip(class_ids, results.masks.xyn):
                flat_coords = " ".join([f"{p[0]:.6f} {p[1]:.6f}" for p in polygon])
                f.write(f"{int(cls_id)} {flat_coords}\n")
                
print("Done! SAM 3 has successfully annotated your dataset.")
```
